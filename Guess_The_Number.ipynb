```python
# Guess the Number

#Project Overview

#This project is a Guess the Number game created using Python.

#The computer will choose a random number between 1 and 10.
#The player must try to guess the number.
#After each guess, the game will tell the people if their guess is too high or too low

```

```python
#imports the random module so a random number can be generated
import random

#Global counters that are used to track the game statisitics
games_played = 0
games_won = 0
games_lost = 0
total_guesses = 0
```

```python
#function containing the main game
def play_game():
    global games_played, games_won, games_lost, total_guesses
    #generates a new secret number for each number
    the_number= random.randint(1,10)

    games_played = games_played + 1

    print("Welcome to Guess the Number!")
    #sets the number of attempts a player can have
    guess = 0
    attempts = 0
    max_attempts = 6
    #keeps asking for guesses until player is correct or out of attempts.
    while guess != the_number and attempts < max_attempts:
        guess = input("Guess a number between 1 and 10: ")
    #checks that the player's input is a number
        if guess.isdigit():
            guess = int(guess)
        else:
            print("Please enter a number.")
            continue

        if guess < 1 or guess > 10:
            print("Please enter a number between 1 and 10.")
            continue

        attempts = attempts + 1
        # gives the player feedback about their guess
        if guess < the_number:
            print("Too low!")
        elif guess > the_number:
            print("Too high!")
        else:
            print("Correct!")
            print("You took", attempts, "attempts.")
            games_won = games_won + 1
    #if the number was not guessed, record a loss
    if guess != the_number:
        print("You lost!")
        print("The correct number was", the_number)
        games_lost = games_lost + 1

    total_guesses = total_guesses + attempts

def game_stats():
    print("Game Statistics")
    print("Games played:", games_played)
    print("Games won:", games_won)
    print("Games lost:", games_lost)
    print("Total guesses:", total_guesses)
```

```python
#starts the game and allows player to replay it
play_again = "yes"
#keep player while the player answers yes
while play_again == "yes":
    play_game()

    #asks the player if they want to play another game
    play_again = input("Would you like to play again? yes/no: ")
    play_again = play_again.lower()

    # checks that player enters either a yes or no
    while play_again != "yes" and play_again != "no":
        print("Please enter yes or no.")
        play_again = input("Would you like to play again? yes/no: ")
        play_again = play_again.lower()

game_stats()
```
